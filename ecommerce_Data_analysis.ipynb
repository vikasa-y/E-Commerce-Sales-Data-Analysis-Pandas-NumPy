## 2. Dataset Overview

After inspecting the dataset, I analyzed **1826 rows and 13 columns** in the working dataset.

The dataset contains information related to:

* Orders

* Revenue
* Payment methods
* Regions
* Order dates

The `order_date` column was in string format converted into a proper datetime format so that year/month-based analysis could be performed.

## 3. Data Quality Analysis

I checked the dataset for common data-quality problems.

### Checks performed

* Missing values
* Duplicate rows
* Data types
* Inconsistent categorical values
* Suspicious numerical values
* Date values

The dataset did not show major data-quality problems in the areas checked.

During date analysis, I noticed that the original dataset contained dates extending beyond 2026. For this analysis, I created a working dataset containing dates up to 2026 while keeping the original dataset unchanged.

---

## 4. Product Category Analysis

I analyzed product categories using order counts and revenue statistics.

### Key findings

* **Electronics** generated the highest total revenue.
* Total revenue from Electronics was approximately **₹18,29,899.22**.
* **Beauty** had the highest average revenue per order, approximately **₹1,059.28**.
* Electronics also had the highest individual revenue value, approximately **₹4,119.33**.
* Beauty had the lowest total revenue among the categories.

### Important observation

The category with the highest total revenue was not necessarily the category with the highest average revenue.

Therefore, "best category" depends on the metric being considered.

**Electronics was the best category based on total revenue, while Beauty was the best based on average revenue per order.**

---

## 5. Payment Method Analysis

I analyzed the different payment methods using order counts and revenue.

### Key findings

* **Card** was the most frequently used payment method.
* **Card** also generated the highest total revenue.
* Therefore, the most frequently used payment method was also the highest-revenue payment method.

### Insight

This indicates that Card was both the most popular payment method and the strongest payment method in terms of total revenue in this dataset.

---

## 6. Regional Analysis

I analyzed order volume and revenue across different regions.

### Key findings

* **West** had the highest number of orders.
* **West** generated the highest total revenue.
* **West** also had the highest average revenue.
* The region with the most orders was also the region with the highest revenue.

### Insight

West was the strongest-performing region across the main metrics analyzed.

---

## 7. Date/Time Analysis

The `order_date` column was converted to datetime and the year was extracted for analysis.

### Key findings

* **2024, 2028, and 2032** had the highest number of orders, with a similar/high order count.
* **2035** had the lowest number of orders.
* **2033** generated the highest total revenue, with revenue exceeding **₹40 lakh**.

### Insight

The year with the highest number of orders was not necessarily the year with the highest total revenue.

This shows that **order volume and revenue are different metrics** and should be analyzed separately.

we use the filtering methods to filter/Remove upcoming years
